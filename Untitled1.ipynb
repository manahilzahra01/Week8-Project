```python
import pandas as pd

# Data load karein
resume_df = pd.read_csv('data/Resume.csv')

print("Data Loaded Successfully!")
print("Total Resumes:", len(resume_df))
print("\nColumns:", resume_df.columns.tolist())
resume_df.head(2)
```

```python
import re
import spacy

nlp = spacy.load('en_core_web_sm')

def clean_text(text):
    if not isinstance(text, str):
        return ""
    # 1. HTML tags removal: Dataset HTML formatted text contain karta hai
    text = re.sub(r'<[^>]+>', ' ', text)
    # 2. Privacy & Noise removal: Email addresses aur URLs remove karna
    text = re.sub(r'\S+@\S+', ' ', text)
    text = re.sub(r'http\S+', ' ', text)
    # 3. Special characters & digits removal: Sirf text content retain karna
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    # 4. Standardization: Lowercase aur extra spaces clean karna
    text = text.lower()
    text = re.sub(r'\s+', ' ', text).strip()
    return text

resume_df['Clean_Resume'] = resume_df['Resume_str'].apply(clean_text)
print("Data Preprocessing Complete!")
resume_df[['Resume_str', 'Clean_Resume']].head(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))
sns.countplot(y='Category', data=resume_df, order=resume_df['Category'].value_counts().index)
plt.title('Resume Distribution Across Job Categories')
plt.xlabel('Number of Samples')
plt.ylabel('Category')
plt.show()

# Word count analysis
resume_df['Word_Count'] = resume_df['Clean_Resume'].apply(lambda x: len(x.split()))
print("Average Word Count per Resume:", resume_df['Word_Count'].mean())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Train/Test Split (80/20 with Random State for Reproducibility)
X_train, X_test, y_train, y_test = train_test_split(
    resume_df['Clean_Resume'], resume_df['Category'], test_size=0.2, random_state=42
)

# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=3000)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# Baseline 1: Classification via Logistic Regression
classifier_model = LogisticRegression(max_iter=1000)
classifier_model.fit(X_train_tfidf, y_train)
y_pred = classifier_model.predict(X_test_tfidf)

print("--- Baseline Classification Report ---")
print(classification_report(y_test, y_pred))
```

```python
from sentence_transformers import SentenceTransformer, util

# Pretrained Sentence-BERT Model Load Karein
sbert_model = SentenceTransformer('all-MiniLM-L6-v2')

# Sample Job Description
sample_jd = "Looking for a Data Scientist proficient in Python, Machine Learning, SQL, and Deep Learning."

# Sentence Embeddings Generate Karein
resume_embeddings = sbert_model.encode(X_test.iloc[:5].tolist(), convert_to_tensor=True)
jd_embedding = sbert_model.encode(sample_jd, convert_to_tensor=True)

# Cosine Similarity Scores Calculate Karein
cosine_scores = util.cos_sim(jd_embedding, resume_embeddings)
print("--- S-BERT Cosine Similarity Scores ---")
print(cosine_scores)
```

```python
def extract_skills_and_gap(resume_text, job_skills):
    # Rule-based skill matching breakdown
    found_skills = [skill for skill in job_skills if skill.lower() in resume_text.lower()]
    missing_skills = [skill for skill in job_skills if skill.lower() not in resume_text.lower()]

    match_score = (len(found_skills) / len(job_skills)) * 100 if job_skills else 0
    return {
        "Match_Score_%": round(match_score, 2),
        "Matched_Skills": found_skills,
        "Missing_Skills": missing_skills
    }

# Test run with sample job skills
sample_job_skills = ['Python', 'SQL', 'Machine Learning', 'Docker', 'AWS', 'NLP']
test_resume = X_test.iloc[0]

result = extract_skills_and_gap(test_resume, sample_job_skills)

print("--- Proposed Improvement: Skill Gap Breakdown ---")
print("Matched Skills:", result['Matched_Skills'])
print("Missing Skills (Gap):", result['Missing_Skills'])
print("Skill Match Score:", result['Match_Score_%'], "%")
```

```python
app_code = """
import streamlit as st
from sentence_transformers import SentenceTransformer, util

st.set_page_config(page_title="AI Resume Matcher", layout="wide")

st.title("🤖 AI-Powered Resume Analyzer & Job Matcher")
st.write("Upload or paste resume text and job requirements to analyze semantic fit and skill gaps.")

col1, col2 = st.columns(2)

with col1:
    jd_input = st.text_area("📄 Job Description:", height=200)
    skills_input = st.text_input("🔑 Required Skills (comma-separated):", "Python, Machine Learning, SQL, AWS")

with col2:
    resume_input = st.text_area("👤 Resume Text:", height=200)

if st.button("🚀 Analyze Match"):
    if jd_input and resume_input:
        with st.spinner("Analyzing semantic similarity..."):
            model = SentenceTransformer('all-MiniLM-L6-v2')
            jd_emb = model.encode(jd_input, convert_to_tensor=True)
            res_emb = model.encode(resume_input, convert_to_tensor=True)
            similarity = util.cos_sim(jd_emb, res_emb).item() * 100

            st.markdown("---")
            st.subheader(f"📊 Overall Semantic Match Score: **{similarity:.2f}%**")

            if skills_input:
                required_skills = [s.strip() for s in skills_input.split(",") if s.strip()]
                matched = [s for s in required_skills if s.lower() in resume_input.lower()]
                missing = [s for s in required_skills if s.lower() not in resume_input.lower()]

                c1, c2 = st.columns(2)
                with c1:
                    st.success(f"✅ **Matched Skills ({len(matched)}):**\n" + (", ".join(matched) if matched else "None"))
                with c2:
                    st.error(f"❌ **Missing Skills Gap ({len(missing)}):**\n" + (", ".join(missing) if missing else "None"))
    else:
        st.warning("⚠️ Please provide both Job Description and Resume Text.")
"""

with open('app/app.py', 'w') as f:
    f.write(app_code)

print("✅ Streamlit app script successfully saved to 'app/app.py'!")
```
